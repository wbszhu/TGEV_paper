```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

data = {'Group':['WT', 'WT', 'PI', 'PI'], 
        'Ratio':[0.72, 0.28, 0.78, 0.22],
        'Type':['Cis', 'Trans', 'Cis', 'Trans']}

df = pd.DataFrame(data) 

df
```

```python
sns.barplot(data=df, y="Ratio", x="Group", hue="Type", palette=("#4F9153","#2A5A63"))
plt.ylim(0,1)
plt.savefig("cis_trans.pdf")
```
