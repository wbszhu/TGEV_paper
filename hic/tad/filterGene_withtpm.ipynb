```python
import pandas as pd

gene = pd.read_table("/public/home/luzhang/08.TGEV/02.results/rna-seq/star_rsem/rsem.merged.gene_tpm.tsv")
gene
```

```python
linkG = pd.read_table("chaLinkG.txt",header=None)
linkG.columns = ["chrom", "start", "end", "geneN", "gene_id", "strand"]
linkG
```

```python
df_out = pd.merge(linkG, gene, on="gene_id")
df_out["TGEV_mean"] = df_out[["TGEV_REP1","TGEV_REP2","TGEV_REP3"]].mean(axis=1)
df_out["WT_mean"] = df_out[["WT_REP1","WT_REP2","WT_REP3"]].mean(axis=1)
df_final = df_out[(df_out["TGEV_mean"]>1)&(df_out["WT_mean"]>1)]
df_final
```

```python
df_final.to_csv("TchaLinkG.txt", index=False, sep="\t")
```
