```python
import pybedtools
bedtool1 = pybedtools.BedTool('100kb_A-B.bed')
bedtool2 = pybedtools.BedTool('gene.bed')
result = bedtool2.intersect(bedtool1, wa=True)
result.saveas('result.bed')
```

```python
!cat result.bed|awk '{print $NF}' |sort -u > 100kb_ABgene_id
!cat result.bed|awk '{print $(NF-1)}'|grep -v "gene_sourceensembl" |sort -u > 100kb_ABgene
```

```python
! wc -l 100kb_ABgene_id
! head  100kb_ABgene_id
```

```python
! wc -l 100kb_ABgene
! head  100kb_ABgene
```
