```python
def plot(file, nam):
    import pandas as pd
    import matplotlib.pyplot as plt
    import seaborn as sns
    import numpy as np
    df = pd.read_table(file, header=None)
    df.columns = ["TPM", "Type"]
    sns.boxplot(x=df["Type"], y = np.log(df["TPM"]),data=df, palette = "Paired")
    plt.legend()
    plt.ylabel("log10(TPM)")
    plt.title(f"{nam} A compartment Vs B compartment")
```

```python
plot("PI_gene_tpm.txt", "PI")
```

```python
plot("WT_gene_tpm.txt", "WT")
```

```python
plot("PI_AorB_gene_tpm.txt", "PI")
```
