```python
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
import matplotlib.patches as patches
import numpy as np
import pandas as pd
```

```python
saddle = np.load('PI.saddle.txt.saddledump.npz', allow_pickle=True)
#x = saddle['saddledata'].shape
up1 = saddle['saddledata'][:10,:10]  # upper-left block
up2 = saddle['saddledata'][42:,:10] # upper-right block
down1 = saddle['saddledata'][:10,42:]  # lower-left block
down2 = saddle['saddledata'][42:,42:] # lower-right block

up1_m = np.mean(up1)
up2_m = np.mean(up2)
down1_m = np.mean(down1)
down2_m = np.mean(down2)

#print(up1_m, up2_m, down1_m, down2_m)
#print(up1.shape, up2.shape, down1.shape, down2.shape)
```

```python
fig, ax = plt.subplots(figsize=(10,10))
norm = LogNorm(vmin=10**(-1), vmax=10**1)
im = ax.imshow(saddle['saddledata'],cmap='bwr',norm = norm)
ax.add_patch(patches.Rectangle((-1, -1),10,10,edgecolor='black',fill=False,lw=1))
ax.add_patch(patches.Rectangle((42, 42),10,10,edgecolor='black',fill=False,lw=1))
ax.add_patch(patches.Rectangle((-1, 42),10,10,edgecolor='black',fill=False,lw=1))
ax.add_patch(patches.Rectangle((42, -1),10,10,edgecolor='black',fill=False,lw=1))
text = ax.text(2.5, 5, "%.2f" % up1_m, fontsize = 20)
text = ax.text(44.5, 5, "%.2f" % up2_m, color = "w", fontsize = 20)
text = ax.text(2.5, 47.5, "%.2f" % down1_m, color = "w", fontsize = 20)
text = ax.text(44.5, 47.5, "%.2f" % down2_m, fontsize = 20)
ax.set_title("Compartmentalization saddle plots for PI", fontsize = 20)
plt.colorbar(im, label='obs/exp', pad=0.025, shrink=0.7)
plt.show()
#plt.savefig("Compartmentalization saddle plots for PI.pdf")
```
