```python
import pybedtools
bedtool1 = pybedtools.BedTool('WT_100kb_A.bed')
bedtool2 = pybedtools.BedTool('gene.bed')
result = bedtool2.intersect(bedtool1, wa=True)
result.saveas('result.bed')
```

```python
!cat result.bed|awk '{print $NF}' |sort -u > WT_100kb_Agene_id
!cat result.bed|awk '{print $(NF-1)}'|grep -v "gene_sourceensembl" |sort -u > WT_100kb_Agene
```

```python
! wc -l WT_100kb_Agene_id
! head WT_100kb_Agene_id
```

```python
! wc -l WT_100kb_Agene
!head WT_100kb_Agene
```
