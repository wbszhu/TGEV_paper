```python
import pandas as pd
```

```python
df = pd.read_csv("ctl-exp.tsv", sep="\t")
df.columns = ["id", "baseMean", "log2FoldChange", "lfcSE", "stat", "pvalue", "padj"]
df
```

```python
df_up = df[(df["log2FoldChange"]>1)&(df["pvalue"]<0.05)]
df_up
```

```python
df_up.set_index("id", inplace = True)
df_up
```

```python
df_down = df[(df["log2FoldChange"]<-1)&(df["pvalue"]<0.05)]
df_down
```

```python
df_down.set_index("id", inplace = True)
df_down
```

```python
df_gene = pd.read_csv("gene.bed", sep="\t")
df_gene.set_index("gene_id", inplace = True)
df_gene
```

```python
df_gene.loc[df_down.index]["GeneSymbol"].to_csv("down_gene.txt", sep="\t")
```

```python
df_gene.loc[df_up.index]["GeneSymbol"].to_csv("up_gene.txt", sep="\t")
```
