```python
import pandas as pd
from tqdm.notebook import tqdm
```

```python
df = pd.read_csv("ctl-exp.tsv", sep="\t")
df.columns = ["id", "baseMean", "log2FoldChange", "lfcSE", "stat", "pvalue", "padj"]
df
```

```python
df_allVar = df[(df["log2FoldChange"].abs()>1)&(df["pvalue"]<0.05)]
df_allVar
```

```python
df_up = df[(df["log2FoldChange"]>1)&(df["pvalue"]<0.05)]
df_top50Up = df_up.sort_values(["log2FoldChange"],ascending = [False]).head(50)
df_top50Up.head()
```

```python
df_down = df[(df["log2FoldChange"]<-1)&(df["pvalue"]<0.05)]
df_top50Down = df_down.sort_values(["log2FoldChange"],ascending = [True]).head(50)
df_top50Down.head()
```

```python
df_top100 = pd.concat([df_top50Up, df_top50Down], axis=0)
```

```python
df_top100.set_index("id", inplace = True)
df_top100
```

```python
df_tpm = pd.read_csv("../star_rsem/rsem.merged.gene_tpm.tsv", sep="\t")
df_tpm.set_index("gene_id", inplace = True)
df_tpm
```

```python
df_tpm.drop(columns=["transcript_id(s)"]).loc[df_top100.index].to_csv("TOP100_variable_tpm.txt", sep='\t', index=True)
```
