# CUT&Tag文件和RNAseq文件（BW）

```python
Mock_RNAseq = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/Mock-RNAseq.bw"
WT_H3K27ac = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/WT-H3K27ac.bw"
WT_H3K4me3 = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/WT-H3K4me3.bw"
WT_H3K27me3 = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/WT-H3K27me3.bw"
WT_promoter = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/WT_promoter.bed"
WT_enhancer = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/WT_enhancer.bed"
```

```python
TGEV_RNAseq = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/TGEV-RNAseq.bw"
PI_H3K27ac = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/PI-H3K27ac.bw"
PI_H3K4me3 = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/PI-H3K4me3.bw"
PI_H3K27me3 = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/PI-H3K27me3.bw"
PI_promoter = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/PI_promoter.bed"
PI_enhancer = "/public/home/jingxu/05.HiTag/01.coolbox/01.TGEV/PI_enhancer.bed"
```

# 整合track

```python
frame1 = XAxis() + \
        Cool(WT_mcool,cmap="RdBu_r",depth_ration=0.5,resolution=10000) + Title("HiC") + TrackHeight(10) + MinValue(-8) + MaxValue(-3) + \
        TADCoverage(WT_tad, border_only=True, alpha=1) + \
        Arcs(WT_loop, line_width=1.5) + Color("#1e1e1e") + Title("Loop") + Inverted() + \
        BED(WT_promoter, alpha=1) + Color("#fc07be") + Title("Promoter") + \
        BED(WT_enhancer, alpha=1) + Color("#fcbe07") + Title("Enhancer") + \
        BigWig(Mock_RNAseq, alpha=0.6) + Color("#91c6b6") + Title("RNAseq") + MinValue(0) + MaxValue(30) + \
        BigWig(WT_H3K4me3, alpha=0.6) + Color("#65503c") + Title("H3K4me3") + MinValue(0) + MaxValue(1400) + \
        BigWig(WT_H3K27ac, alpha=0.6) + Color("#e18d73") + Title("H3K27ac") + MinValue(0) + MaxValue(350) + \
        BigWig(WT_H3K27me3, alpha=0.6) + Color("#747ea3") + Title("H3K27me3") + MinValue(0) + MaxValue(20) + \
        BED(pig101) + Title("GTF Annotation")
frame1.plot(region)

```

```python
frame2 = XAxis() + \
        Cool(PI_mcool,cmap="RdBu_r",depth_ration=0.5,resolution=10000) + Title("HiC") + TrackHeight(10) + MinValue(-8) + MaxValue(-3) + \
        TADCoverage(PI_tad, border_only=True, alpha=1) + \
        Arcs(PI_loop, line_width=1.5) + Color("#1e1e1e") + Title("Loop") + Inverted() + \
        BED(PI_enhancer, alpha=1) + Color("#fcbe07") + Title("Enhancer") + \
        BigWig(TGEV_RNAseq, alpha=0.6) + Color("#91c6b6") + Title("RNAseq") + MinValue(0) + MaxValue(30) + \
        BigWig(PI_H3K4me3, alpha=0.6) + Color("#65503c") + Title("H3K4me3") + MinValue(0) + MaxValue(1400) + \
        BigWig(PI_H3K27ac, alpha=0.6) + Color("#e18d73") + Title("H3K27ac") + MinValue(0) + MaxValue(350) + \
        BigWig(PI_H3K27me3, alpha=0.6) + Color("#747ea3") + Title("H3K27me3") + MinValue(0) + MaxValue(20) + \
        GTF(pig101) + Title("GTF Annotation")
frame2.plot(region)
```
