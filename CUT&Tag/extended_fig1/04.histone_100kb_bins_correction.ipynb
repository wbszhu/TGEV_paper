```python
import numpy as np
import pandas as pd
```

```python
#read multibam histone signal
df = pd.read_table("/path/to/cutandtag/data/scores_per_100kb_new.tab")
df.columns = ["chr", "start", "end", "wt-k27ac_1", "wt-k27ac_2",'wt-4me3_1','wt-4me3_2','wt-27me3_1','wt-27me3_2', "wt-input_1", "wt-input_2", "pi-k27ac_1", "pi-k27ac_2",'pi-4me3_1','pi-4me3_2','pi-27me3_1','pi-27me3_2',"pi-input_1", "pi-input_2"]
df.head()
```

```python
#h3k4me3_R1,h3k4me3_R2
#41,285,688 38,983,472
```

### normalization of sequence depth

```python
#calibrated depth
df["wt-k27ac_1nor"] = df["wt-k27ac_1"]/47
df["wt-k27ac_2nor"] = df["wt-k27ac_2"]/53
df["wt-4me3_1nor"] = df["wt-4me3_1"]/37
df["wt-4me3_2nor"] = df["wt-4me3_2"]/34
df["wt-k27me3_1nor"] = df["wt-27me3_1"]/63
df["wt-k27me3_2nor"] = df["wt-27me3_2"]/73
df["wt-igg_1nor"] = df["wt-input_1"]/10
df["wt-igg_2nor"] = df["wt-input_2"]/10
df["pi-k27ac_1nor"] = df["pi-k27ac_1"]/38
df["pi-k27ac_2nor"] = df["pi-k27ac_2"]/59
df["pi-4me3_1nor"] = df["pi-4me3_1"]/13
df["pi-4me3_2nor"] = df["pi-4me3_2"]/58
df["pi-k27me3_1nor"] = df["pi-27me3_1"]/26
df["pi-k27me3_2nor"] = df["pi-27me3_2"]/46
df["pi-igg_1nor"] = df["pi-input_1"]/8
df["pi-igg_2nor"] = df["pi-input_2"]/9
df.head()
```

```python
#compute average signal
df['wt-k27ac_mean'] = df[['wt-k27ac_1nor', 'wt-k27ac_2nor']].mean(axis=1)
df['wt-4me3_mean'] = df[['wt-4me3_1nor', 'wt-4me3_2nor']].mean(axis=1)
df['wt-k27me3_mean'] = df[['wt-k27me3_1nor', 'wt-k27me3_2nor']].mean(axis=1)
df['wt-igg_mean'] = df[['wt-igg_1nor', 'wt-igg_2nor']].mean(axis=1)
```

```python
df['pi-k27ac_mean'] = df[['pi-k27ac_1nor', 'pi-k27ac_2nor']].mean(axis=1)
df['pi-4me3_mean'] = df[['pi-4me3_1nor', 'pi-4me3_2nor']].mean(axis=1)
df['pi-k27me3_mean'] = df[['pi-k27me3_1nor', 'pi-k27me3_2nor']].mean(axis=1)
df['pi-igg_mean'] = df[['pi-igg_1nor', 'pi-igg_2nor']].mean(axis=1)
```

```python
df.head()
```

```python
#remove control background
df["wt-k27ac_change"] = df["wt-k27ac_mean"]-df["wt-igg_mean"]
df["wt-4me3_change"] = df["wt-4me3_mean"]-df["wt-igg_mean"]
df["wt-k27me3_change"] = df["wt-k27me3_mean"]-df["wt-igg_mean"]
df["pi-k27ac_change"] = df["pi-k27ac_mean"]-df["pi-igg_mean"]
df["pi-4me3_change"] = df["pi-4me3_mean"]-df["pi-igg_mean"]
df["pi-k27me3_change"] = df["pi-k27me3_mean"]-df["pi-igg_mean"]
df.head()
```

```python
#save h3k27ac result
h3k27ac = df[["chr","start","end","wt-k27ac_change","pi-k27ac_change"]] 
h3k27ac.head()
h3k27ac.to_csv("h3k27ac_change.tab", sep="\t", index=False)
```

```python
#save h3k4me3 result
h3k4me3 = df[["chr","start","end","wt-4me3_change","pi-4me3_change"]] 
h3k4me3.head()
h3k4me3.to_csv("h3k4me3_change.tab", sep="\t", index=False)
```

```python
#save h3k27me3 result
h3k27me3 = df[["chr","start","end","wt-k27me3_change","pi-k27me3_change"]] 
h3k27me3.head()
h3k27me3.to_csv("h3k27me3_change.tab", sep="\t", index=False)
```

```python
#remove noise
h3k4me3 = pd.read_table("h3k4me3_change.tab")
h3k4me3.shape
```

```python
h3k4me3[['wt-4me3_change', 'pi-4me3_change']] = h3k4me3[['wt-4me3_change', 'pi-4me3_change']].applymap(lambda x: 1 if x < 0 else x)
h3k4me3[['wt-4me3_change', 'pi-4me3_change']] = h3k4me3[['wt-4me3_change', 'pi-4me3_change']].apply(np.log)
h3k4me3.to_csv("h3k4me3_log.txt", sep="\t", index=False)
```
