```python
import pandas as pd 
```

```python
df = pd.read_table("/path/to/cutandtag/data/101TGEV/03.vis/03.deg_slm/random_peakReadcount.txt")
df.columns = ["chr", "start", "end", "wt-k27ac_1", "wt-k27ac_2", "wt-input_1", "wt-input_2", "pi-k27ac_1", "pi-k27ac_2", "pi-input_1", "pi-input_2"]
```

```python
#h3k4me3_R1,h3k4me3_R2
#41,285,688 38,983,472
```

### normalization of sequence depth

```python
df["wt-k27ac_1nor"] = df["wt-k27ac_1"]/47
df["wt-k27ac_2nor"] = df["wt-k27ac_2"]/53
df["wt-igg_1nor"] = df["wt-input_1"]/10
df["wt-igg_2nor"] = df["wt-input_2"]/10
df["pi-k27ac_1nor"] = df["pi-k27ac_1"]/38
df["pi-k27ac_2nor"] = df["pi-k27ac_2"]/59
df["pi-igg_1nor"] = df["pi-input_1"]/8
df["pi-igg_2nor"] = df["pi-input_2"]/9
df
```

```python
def mean_of_cols(row):
        return row[["wt-k27ac_1nor","wt-k27ac_2nor"]].mean()

    

df["wt-k27ac_mean"] = df.apply(mean_of_cols, axis=1)
```

```python
def mean_of_cols(row):
        return row[["pi-k27ac_1nor","pi-k27ac_2nor"]].mean()

    

df["pi-k27ac_mean"] = df.apply(mean_of_cols, axis=1)
```

```python
def mean_of_cols(row):
        return row[["wt-igg_1nor","wt-igg_2nor"]].mean()

    

df["wt-igg_mean"] = df.apply(mean_of_cols, axis=1)
```

```python
def mean_of_cols(row):
        return row[["pi-igg_1nor","pi-igg_2nor"]].mean()

    

df["pi-igg_mean"] = df.apply(mean_of_cols, axis=1)
```

```python
df
```

```python
df["wt-k27ac_foldchange"] = df["wt-k27ac_mean"]/df["wt-igg_mean"]
df["pi-k27ac_foldchange"] = df["pi-k27ac_mean"]/df["pi-igg_mean"]
df
```

```python
col_list = ["wt-k27ac_foldchange", "pi-k27ac_foldchange"]
df.to_csv("random_foldchange.bed", columns=col_list, sep="\t", index=False, header=True)
```
