# CUT&Tag文件和RNAseq文件（BW）

```python
#Mock_RNAseq = "/public/home/jingxu/05.HiTag/01.coolbox/00.data/Mock-RNAseq.bw"
TGEV_RNAseq = "/public/home/jingxu/05.HiTag/01.coolbox/00.data/TGEV-RNAseq.bw"
#WT_H3K27ac = "/public/home/jingxu/05.HiTag/01.coolbox/00.data/WT-H3K27ac.bw"
PI_H3K27ac = "/public/home/jingxu/05.HiTag/01.coolbox/00.data/PI-H3K27ac.bw"
#WT_H3K4me3 = "/public/home/jingxu/05.HiTag/01.coolbox/00.data/WT-H3K4me3.bw"
PI_H3K4me3 = "/public/home/jingxu/05.HiTag/01.coolbox/00.data/PI-H3K4me3.bw"
#WT_H3K27me3 = "/public/home/jingxu/05.HiTag/01.coolbox/00.data/WT-H3K27me3.bw"
PI_H3K27me3 = "/public/home/jingxu/05.HiTag/01.coolbox/00.data/PI-H3K27me3.bw"
```

# 整合track

```python
frame = XAxis() + \
        Cool(PI_mcool,cmap="RdYlBu_r",depth_ration=0.5,resolution=10000) + Title("HiC") + TrackHeight(8) + \
        BigWig(TGEV_RNAseq, alpha=0.6) + Color("#5083bb") + Title("RNAseq") + MinValue(0) + MaxValue(1) + \
        BigWig(PI_H3K27ac, alpha=0.6) + Color("#b6deec") + Title("H3K27ac") + MinValue(0) + MaxValue(20) + \
        BigWig(PI_H3K4me3, alpha=0.6) + Color("#fdd081") + Title("H3K4me3") + MinValue(0) + MaxValue(20) + \
        BigWig(PI_H3K27me3, alpha=0.6) + Color("#ce2726") + Title("H3K27me3") + MinValue(0) + MaxValue(20) + \
        GTF(pig101) + Title("GTF Annotation")
frame.plot(region)
```

```python
bsr = Browser(frame)
bsr.goto(region)
bsr.show()
```

```python
bsr.save("PI-coolbox.pdf")
```
