# Load coolbox

```python
import coolbox
from coolbox.api import *
```

# Check coolbox version

```python
coolbox.__version__
```

# Target region: upstream and downstream of ANPEP

```python
region = "chr7:55250000-55450000"
```

# Reference genome file

```python
pig101 = "/path/to/reference/Sscrofa101/Sus_scrofa.Sscrofa11.1.101.chr.gtf"
```

# Hi-C file

```python
#WT_mcool = "/path/to/hic/data/01.coolbox/00.data/WT.balanceNoMTY.mcool"
PI_mcool = "/path/to/hic/data/01.coolbox/00.data/PI.balanceNoMTY.mcool"
```

# CUT&Tag and RNA-seq files (BigWig)

```python
#Mock_RNAseq = "/path/to/hic/data/01.coolbox/00.data/Mock-RNAseq.bw"
TGEV_RNAseq = "/path/to/hic/data/01.coolbox/00.data/TGEV-RNAseq.bw"
#WT_H3K27ac = "/path/to/hic/data/01.coolbox/00.data/WT-H3K27ac.bw"
PI_H3K27ac = "/path/to/hic/data/01.coolbox/00.data/PI-H3K27ac.bw"
#WT_H3K4me3 = "/path/to/hic/data/01.coolbox/00.data/WT-H3K4me3.bw"
PI_H3K4me3 = "/path/to/hic/data/01.coolbox/00.data/PI-H3K4me3.bw"
#WT_H3K27me3 = "/path/to/hic/data/01.coolbox/00.data/WT-H3K27me3.bw"
PI_H3K27me3 = "/path/to/hic/data/01.coolbox/00.data/PI-H3K27me3.bw"
```

# Assemble tracks

```python
frame = XAxis() + \
        Cool(PI_mcool,cmap="RdYlBu_r",depth_ration=0.5,resolution=10000) + Title("HiC") + TrackHeight(8) + \
        BigWig(TGEV_RNAseq, alpha=0.6) + Color("#5083bb") + Title("RNAseq") + MinValue(0) + MaxValue(1) + \
        BigWig(PI_H3K27ac, alpha=0.6) + Color("#b6deec") + Title("H3K27ac") + MinValue(0) + MaxValue(20) + \
        BigWig(PI_H3K4me3, alpha=0.6) + Color("#fdd081") + Title("H3K4me3") + MinValue(0) + MaxValue(20) + \
        BigWig(PI_H3K27me3, alpha=0.6) + Color("#ce2726") + Title("H3K27me3") + MinValue(0) + MaxValue(20) + \
        GTF(pig101) + Title("GTF Annotation")
frame.plot(region)
```

```python
bsr = Browser(frame)
bsr.goto(region)
bsr.show()
```

```python
bsr.save("PI-coolbox.pdf")
```
