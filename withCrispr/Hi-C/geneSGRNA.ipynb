```python
import pandas as pd
import numpy as np
df = pd.read_csv("all_sgRNA.txt", sep="\t")
df.columns = ["sgRNA","GeneID","PigGeCKO cells_read1","PigGeCKO cells_read2","PigGeCKO cells_read1_norm","PigGeCKO cells_read2_norm","average(PigGeCKO cells)","read1","read2","read1_norm","read2_norm","average","Fold_change","log2(Fold_change)"]
df['GeneID'] = df['GeneID'].str.lstrip(" ")
df
```

```python
df = df[df['log2(Fold_change)'].str.match("-?\d+.*")]
df['log2(Fold_change)'] = df['log2(Fold_change)'].astype(np.float64)
```

```python
df[df["GeneID"] == "ANPEP"]
```

```python
idx = df.groupby(['GeneID'])['log2(Fold_change)'].transform(max) == df['log2(Fold_change)']
```

```python
df_test = df[idx]
df_test
```
