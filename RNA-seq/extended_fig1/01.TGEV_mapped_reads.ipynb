```python
import matplotlib as mpl
mpl.rcParams['pdf.fonttype'] = 42
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#reads percentage
data = {'Group':['Mock', 'TGEV'], 
        'Ratio':[0.002, 33.830]}

df = pd.DataFrame(data) 

df
```

```python
#plot
fig, ax = plt.subplots(figsize=(3, 4))
sns.barplot(data=df, y="Ratio", x="Group", width= 0.3, palette=("#304172","#c0271d"))
plt.ylim(0,100)
plt.ylabel('Percentage of viral reads in RNA-Seq')
plt.title('TGEV reads of RNA-Seq')
plt.savefig("/path/to/rnaseq/data/tgev_raeds_ratio.pdf", bbox_inches='tight')
plt.tight_layout()
plt.show()
```
